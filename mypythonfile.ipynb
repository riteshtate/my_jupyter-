# What are Identifiers?

Rules for Naming Identifiers


1.	Can contain letters (A-Z, a-z), digits (0-9), and underscores (_) only
o	✅ my_variable, data123, _hidden_value
o	❌ 123var, my-variable, @name (Invalid)

2.	Must not start with a digit
o	✅ variable1, _myvar
o	❌ 1variable, 123abc (Invalid)

3.	Cannot be a Python keyword
o	❌ if, else, class, return, lambda, etc. (Reserved words)

4.	Case-sensitive
o	✅ MyVariable and myvariable are different

5.	No special characters allowed (except _)
o	❌ var@name, my#value (Invalid)

## Types of Operators in Python

### 1. Arithmetic Operators

👉 Used for mathematical calculations.

| Symbol | Name           | Description           | IP        | OP    |
| ------ | -------------- | --------------------- | --------- | ----- |

### 2. Comparison Operators

## Loop Control Statements

| Statement  | Purpose                 |
| ---------- | ----------------------- | 
| `break`    | Stops loop completely   | 
| `continue` | Skips current iteration | 
| `pass`     | Does nothing            | 


1. break Statement

👉 break immediately stops the loop completely.
👉 Once break executes, the loop terminates.

```python
for i in range(1,6):   # 1,2,3
    
    if i==3:            #1==3 false
        break
    print(i)             # 1,2
    
```

2. continue Statement

👉 continue skips the current iteration and moves to the next iteration.
👉 Loop does NOT stop.

```python
for i in range(1,6):   # 1,2,3
    
    if i==3:            #3==3 true  
        continue
    print(i)             # 1,2,4,5
    
```

3. pass Statement

👉 pass is a placeholder statement.
👉 It does nothing.
👉 Used when code will be added later.


    Why pass is Used?

✅ Avoids syntax error
✅ Used during future planning
✅ Creates empty loops/functions/classes temporarily

```python
for i in range(1,11):
    pass
```

# 2. while Loop


👉 A while loop executes a block of code repeatedly as long as the condition is True.
👉 It is called a condition-based loop.

Syntax 


while condition:
    statement

How while Loop Works?

Condition is checked
If condition is True → code executes
Again condition is checked
Repeats until condition becomes False

```python
# 1,.........,10

i = 1                    # 1 

while i<11:           # 1<11
    print(i)            # 1
    
    i=i+1                  #2                                                           # i+=1 this format use mainly time
```

```python
# 2 ,4,6,................,20

i = 2

while i<21:              # 2<21 true
    print(i)             # 2
    
    i=i+2                 # i=2+2
```

## Loop Control Statements in while Loop

```python
# break

i = 1                    

while i<6:    # 1<6  true
     
    if i==3:        # 1==3   false
        break
        
    print(i)      #  1,2
   
    i=i+1          # 1+1 = 2    # i+=1 this format use mainly time
```

```python
i = 1                    

while i<6:     # 
     
    if i==3:
        continue
    
    print(i)      
    i=i+1       
```
