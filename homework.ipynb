#### Arithmetic operator

#### +     -      *      /        %        **        //

```python
a=20
b=30

print(a+b)  # addition (+)   operator   20+30

print(a-b)  # subtraction (-)  operator  20-30

print(a*b)  # multiplication (*)  operator 20*30

print(b/a)  # division (/) operator 30/20
print("\n")
print(a%b)  # modulus (%) operator a%b
print("\n")
print(a**b) # Exponent (**) operator  a power b
print("\n")
print(b//a) # float division operator (//) are removed decimal value   a//b
```

#### comparison operator
#### ==,!=,>,<,>=,<=

```python
a = 5
b = 60
c = 5

print(a==b)  # 5 == 60   check euality
print(a==c)  # 5 == 5    

print("\n")

print(a != b)  # 5 != 60  check ineuality

print("\n")

print(a<=b)  # 5<60     check lessthan euality
print(a>=b)  # 5>60     check greater than euality
```

#### Logical Operator 
#### AND ---- any false ----- false
#### OR  ---- any true  ----- true
#### not ----  true  -- false -- true 

```python
a = True
b = False

print(a and b)   # a and b = false
print(a or b)    # a or b = true 
print(not a)     # not a = false
```

## if - else  statement

```python
# check positive, negative 
num = -20
if num >= 0:
    print("positive")
    
else:
    print("negative")
    
print("\n")
#check even ,odd
num = 33
if num % 2==0:
    print("even")
    
else:
    print("odd")
    
print("\n")
# check voting eligibal or not
age = 17
if age>=18:
    print("eligibal for voting")
    
else:
    print("not eligible for voting")

print("\n")
#check pass or fail
marks=45
if marks >= 35:
    print("pass")
    
else:
    print("fail")
    
print("\n")
# check greater number
a=20
b=45

if a>b:
    print("greater")
    
else:
    print("less")
```
